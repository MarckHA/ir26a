Ejemplo para mostrar el tiempo de ejecución de la búsqueda básica

```python
import time

start_time = time.time()

query = "Mary"
encontrados = []
for key in contenidos.keys():
    if query in contenidos[key]:
        encontrados.append(os.path.basename(key))

print(f"La palabra se encuentra en los siguientes archivos:\n {encontrados}") 

end_time = time.time()
print(f"Tiempo de ejecución: {end_time - start_time} en segundos")
```

#### Definición de la función buscar_palabra

```python
def buscar_palabra(query, contenidos):
    encontrados = []
    for key in contenidos.keys():
        if query in contenidos[key]:
            encontrados.append(key)
    return encontrados
```
