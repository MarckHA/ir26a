## Ejercicio 02 - Índice Invertido
Un índice invertido es una estructura de datos en la que se detalla en qué documentos aparecen cada una de las palabras del vocabulario dentro del corpus

```python
import os
path = "C:\\Users\\mark_\\Documents\\ir26a\\Tareas\\Tarea01\\data"
files = os.listdir(path)
len(files)
```

```python
contenidos = {}
archivo = files[0]
indice_invertido = {}

with open(path + "\\" + archivo, "r", encoding="utf-8") as f:
    contenido = f.read()
    palabras = contenido.split()
    vocabulario = set(palabras)
    print(f"20 primera palabras dentro del archivo {archivo}: \n", palabras[:20])
    print(f"Cantidad de palabras en el archivo {archivo}: ",len(palabras))
    print(f"Cantidad de palabras únicas en el archivo {archivo}: ",len(vocabulario))
```

Creamos los índices invertidos para nuestro vocabulario del texto. La impresión muestra las 20 primeras palabras únicas y el libro en el que se encuentran. Al evaluar un solo libro es el único que sale.

```python
for palabra in vocabulario:
    indice_invertido[palabra] = [archivo]

for i, (clave, valor) in enumerate(indice_invertido.items()):
    if i == 20:
        break
    print(clave, ":", valor)
```

Para dos archivos:

```python
contenidos = {}
indice_invertido = {}

for archivo in files[:2]:
    # Se iteran los archivos
    with open(path + "\\" + archivo, "r", encoding="utf-8") as f:
        contenido = f.read()
        palabras = contenido.split()
        vocabulario = set(palabras)
        for palabra in vocabulario:
            if palabra in indice_invertido:
                indice_invertido[palabra].append(archivo)
            else:
                indice_invertido[palabra] = [archivo]
        print('Archivo:', archivo, '    - Número de palabras únicas:', len(vocabulario))
        print('Número de palabras del índice invertido: ',len(indice_invertido))
```

Se utilizará una librería para poder visualizar de mejor manera los resultados del diccionario indice_invertido

### Función buscar 
Retorna la lista de los archivos en los que la palabra aparece

```python
def buscar_palabra(palabra):
    return indice_invertido.get(palabra, "La palabra no se encuentra en ningún texto.")
```

Aplicamos la función buscar para probar nuestra función entre estos dos libros, se muestran los primeros 20 resultados.

```python
for i, palabra in enumerate(indice_invertido.keys()):
    if i == 20:
        break
    print(f"{palabra}: {buscar_palabra(palabra)}")
```

Para todos los archivos:

```python
contenidos = {}
indice_invertido = {}

for archivo in files:
    # Se iteran los archivos
    with open(path + "\\" + archivo, "r", encoding="utf-8") as f:
        contenido = f.read()
        palabras = contenido.split()
        vocabulario = set(palabras)
        for palabra in vocabulario:
            if palabra in indice_invertido:
                indice_invertido[palabra].append(archivo)
            else:
                indice_invertido[palabra] = [archivo]
        print('Archivo:', archivo, '    - Número de palabras únicas:', len(vocabulario))
        print('Número de palabras del índice invertido: ',len(indice_invertido))
```

Se muestran los primeros 20 resultados de la búsqueda de la palabra en los 100 libros:

```python
for i, palabra in enumerate(indice_invertido.keys()):
    if i == 20:
        break
    print(f"{palabra}: {buscar_palabra(palabra)}")
```

Verificamos el tiempo que le toma realizar esta búsqueda entre los 100 libros

```python
import time
start_time = time.time()
for palabra in indice_invertido.keys():
    buscar_palabra(palabra)
end_time = time.time()
print(f"Tiempo de ejecución: {end_time - start_time} en segundos")
```
